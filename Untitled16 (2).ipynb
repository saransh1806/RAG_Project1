```python
def answer_question(
    query,
    context=None,
    top_k_semantic_and_keyword=50,
    top_k_from_combined_semantic_and_keyword=20,
    final_top_k=5,
    K=60,
    is_generate_summary=False
):
    """
    Processes one user question using the existing RAG pipeline.
    """

    if context is None:
        context = []

    # Add the current question to conversation history
    context.append({
        "role": "user",
        "content": query
    })

    # Generate answer using the existing RAG pipeline
    response = chatbot.generate(
        query,
        chatbot.model,
        chatbot.tokenizer,
        chatbot.model,
        chatbot.tokenizer,
        chatbot.embedding_model,
        context,
        chatbot.collection,
        chatbot.reranker,
        top_k_semantic_and_keyword=top_k_semantic_and_keyword,
        top_k_from_combined_semantic_and_keyword=top_k_from_combined_semantic_and_keyword,
        final_top_k=final_top_k,
        K=K,
        is_generate_summary=is_generate_summary
    )

    # Store the assistant response
    context.append({
        "role": "assistant",
        "content": response
    })

    return response

from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, Field

app = FastAPI(
    title="MIT 6.006 RAG Chatbot API",
    description="API for an educational RAG chatbot",
    version="1.0.0"
)
from fastapi.middleware.cors import CORSMiddleware

app.add_middleware(
    CORSMiddleware,
    allow_origins=[
        "http://127.0.0.1:5500",
        "http://localhost:5500"
    ],
    allow_credentials=False,
    allow_methods=["*"],
    allow_headers=["*"],
)


class ChatRequest(BaseModel):
    question: str = Field(
        min_length=1,
        description="Question asked by the user"
    )
    history: list[dict] = Field(default_factory=list)


class ChatResponse(BaseModel):
    answer: str
    sources: list = []


@app.get("/")
def home():
    return {
        "message": "MIT 6.006 RAG API is running"
    }


@app.get("/health")
def health():
    return {
        "status": "healthy"
    }


@app.post("/chat", response_model=ChatResponse)
def chat_endpoint(request: ChatRequest):

    try:
        # We will connect your existing RAG function here
        answer = answer_question(
            query=request.question,
            context=request.history
        )
        return ChatResponse(answer=answer)
    except Exception as e:
        raise HTTPException(
            status_code=500,
            detail=str(e)
        )
```

```python
# 1. Run FastAPI in a background thread
import threading, uvicorn, nest_asyncio
nest_asyncio.apply()

threading.Thread(
    target=lambda: uvicorn.run(app, host="0.0.0.0", port=8000),
    daemon=True
).start()
```
